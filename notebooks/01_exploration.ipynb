## Structure and data quality

```python
print(df.shape)
df.info()
```

```python
# Missing values per column (count and %)
missing = df.isna().sum().to_frame("missing")
missing["pct"] = (missing["missing"] / len(df) * 100).round(1)
missing[missing["missing"] > 0].sort_values("pct", ascending=False)
```

```python
# Duplicates and summary statistics
print("Duplicate rows:", df.duplicated().sum())
df.describe(include="all").T
```
